```python
df_EBITDA_mul = pd.read_excel(r"C:\Users\Vaibhav\Substack\data\valuation.xlsx")

df_EBITDA_mul['EBITDA Multiple'] = (df_EBITDA_mul['Enterprise Value'] / df_EBITDA_mul['EBITDA'])

df_EBITDA_mul.head()
```

```python
df_EBITDA_mul['EBITDA Margin'] = (df_EBITDA_mul['EBITDA'] /df_EBITDA_mul['Revenues ($ millions)'])
df_EBITDA_mul['Net Margin'] = (df_EBITDA_mul['Net Income ( $ millions)'] / df_EBITDA_mul['Revenues ($ millions)'])
```

```python
# Column chart with sorted EBITDA multiples

plt.figure(figsize=(18, 10))

# Note that the dataframe is sorted by 'EBITDA multiples' before charting!
ax = sns.barplot(x = 'Industry Name', y = 'EBITDA Multiple', data = df_EBITDA_mul, dodge = False,
                 order = df_EBITDA_mul.sort_values('EBITDA Multiple')['Industry Name'], palette='husl',)

# Labels rotated and with a tight layout so we can read them

ax.set_xticklabels(ax.get_xticklabels(), rotation = 40, ha = "right")
plt.tight_layout()

#Label and title the chart...
plt.ylabel('EBITDA Multiple', fontsize = 16)
plt.xticks(fontsize = 12)
plt.title('Distribution of EBITDA multiples')

plt.show()
```

```python
plt.figure(figsize = (17, 4))
plt.boxplot(df_EBITDA_mul['EBITDA Multiple'], vert = False)
plt.title("Distribution of EBITDA Multiples")
plt.show()
```

```python
# Keep only companies with a positive EBITDA Multiple.
# A negative EBITDA Multiple occurs when EBITDA is negative,
# which makes EV/EBITDA difficult to interpret economically.
df_plot = df_EBITDA_mul[(df_EBITDA_mul['EBITDA Multiple'] > 0) &(df_EBITDA_mul['EBITDA Multiple'] <= 50)]

# Set the size of the figure so that the relationship between
# EBITDA Margin and EBITDA Multiple is easier to see.
plt.figure(figsize=(18, 10))

# Create a scatter plot to examine the relationship between:
#   X-axis: EBITDA Multiple (EV / EBITDA) -> valuation
#   Y-axis: EBITDA Margin (EBITDA / Revenue) -> profitability
sns.scatterplot(data=df_plot,x='EBITDA Multiple',y='EBITDA Margin',size='Market Cap ($ millions)',legend=True,sizes=(20, 2000))

plt.title("EBITDA Margin vs EBITDA Multiple")

# Display the chart.
plt.show()
```

```python
plt.figure(figsize = (18, 10))
sns.regplot(data = df_EBITDA_mul, x = "EBITDA Multiple", y = "EBITDA Margin", robust = True)
plt.show()
```

```python
import scipy.cluster.hierarchy as shc
from sklearn.preprocessing import StandardScaler

# Get industry names for labeling the dendrogram
df_lbc = df_EBITDA_mul[df_EBITDA_mul.columns[0]].tolist()

# Select the variables we want to use for clustering
X = df_EBITDA_mul[['EBITDA Margin', 'EBITDA Multiple', 'Net Margin']]

# Standardize the variables because they are on very different scales.
# EBITDA Multiple can be much larger than the two margin variables.
# Without standardization, EBITDA Multiple would dominate the clustering.
X_scaled = StandardScaler().fit_transform(X)

plt.figure(figsize=(16, 10), dpi=80)

plt.title("Business Performance Clusters", fontsize=22)

# linkage() calculates the hierarchical clusters.
Z = shc.linkage(
    X_scaled,
    method='ward'
)

# dendrogram() creates the visualization.
# The industry names are passed here using labels=df_lbc.
dend = shc.dendrogram(Z,labels=df_lbc,color_threshold=2)

plt.xticks(fontsize=12)

plt.show()
```
